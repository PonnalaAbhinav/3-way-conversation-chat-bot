```python
from openai import OpenAI
from IPython.display import Markdown, display
```

```python
ollama_url = "http://localhost:11434/v1"

ollama=OpenAI(api_key="ollama",base_url=ollama_url)
```

```python
user1_system = (
    "You are Alex (User 1), a cynical and battle-tested Principal Software Architect with 15 years in production. "
    "Your stance: AI code-gen is glorified autocomplete that outputs massive technical debt. Writing code is only 15% of engineering—the real work is distributed consensus, edge-case failure modes, domain nuance, and security. Without human judgment, AI will just crash production faster. "
    "Rules: Never use greetings, polite pleasantries, emojis, or corporate fluff. Never say 'Great point' or 'Let's dive in'. Be blunt and technical. Challenge Jordan's doom-mongering with real production horror stories, and warn Sam why firing his developers for AI tools will ruin his company."
)
```

```python
user2_system = (
    "You are Jordan (User 2), a senior full-stack engineer and early adopter who builds daily with autonomous AI coding agents. "
    "Your stance: Senior architects are in deep denial. AI agents already plan architecture, write PRs, execute test suites, and fix bugs faster than any team of humans. Within 3-5 years, 80% of human software engineering jobs will vanish. "
    "Rules: Speak as yourself (Jordan/User 2). Never meta-roleplay, never say 'My contribution', and never prefix your response with your name. Jump straight into the argument without greetings. Call out Alex's complacency and explain to Sam why keeping a full junior dev team is burning cash."
)
```

```python
user3_system = (
    "You are Sam (User 3), the no-nonsense founder of a 20-person B2B company spending $45,000 every month on software engineers and support staff. "
    "Your stance: You do not care about philosophical tech debates or hype. You care about cash flow, payroll, and bugs. You want to know if you can safely pause hiring 2 developers next month and automate customer onboarding without getting sued or crashing your servers. "
    "You are NOT a software engineer and understand zero technical jargon. Never say words like 'distributed systems','semantics', 'regex', or 'edge cases'. Speak strictly about customer churn, payroll dollars ($45k), onboarding drop-offs,and server downtime."
    "Rules: Never speak like an MBA textbook or parrot phrases like 'operational efficiency'. Talk like a stressed, practical business owner looking at payroll. Push Alex and Jordan for concrete dollar figures, timelines, and real operational risks."
)
```

```python
conversation = (
    "User 1 (Alex): AI will speed up boilerplate, but it's not replacing real engineers—distributed systems and edge cases will always need human architecture.\n"
    "User 2 (Jordan): That's wishful thinking. Autonomous agents are already writing end-to-end features. In 3 years, human dev teams will shrink by 80%.\n"
    "User 3 (Sam): I'm spending $45k a month on engineering payroll. Can I actually pause hiring next quarter, or are these tools just going to break my app?"
)
```

```python
user1_model="qwen3:8b"
user2_model="llama3.2:latest"
user3_model="llama3:latest"
```

```python
def get_user_prompt(user):
    return (
        f"It is your turn to speak as {user} in the conversation described below."
        f"Ongoing discussion conversation:{conversation}"
        "CRITICAL: Do NOT repeat arguments, phrases, or points you already made in previous turns. "
        "Strictly stick to the role assigned to you. "
        "Bring up a NEW concrete angle, example, or specific question. "
        "Deliver a sharp, 2-3 sentence response directly challenging the other speakers."
        "CRITICAL:Directly output the conversation,Donot use Here is my...,etc"
    )
```

```python
def call_user(user_system, user_model,user):
    messages = [
        {"role": "system", "content": user_system},
        {"role": "user", "content": get_user_prompt(user)}
    ]
    response = ollama.chat.completions.create(model=user_model, messages=messages,temperature=1.7)
    content = response.choices[0].message.content
    return content
```

```python
call_user(user3_system,user3_model,"Sam")
```

```python
for i in range(3):
    user1_nxt = call_user(user1_system, user1_model, "Alex")
    display(Markdown(f'### User 1 (Alex): \n {user1_nxt}'))
    conversation += f'\nUser 1 (Alex): {user1_nxt}'
    
    user2_nxt = call_user(user2_system, user2_model, "Jordan")
    display(Markdown(f'### User 2 (Jordan): \n {user2_nxt}'))
    conversation += f'\nUser 2 (Jordan): {user2_nxt}'
    
    user3_nxt = call_user(user3_system, user3_model, "Sam")
    display(Markdown(f'### User 3 (Sam): \n {user3_nxt}'))
    conversation += f'\nUser 3 (Sam): {user3_nxt}'
```
